```python
data=pd.merge(new_meta,credits,on='id')
```

```python
pd.set_option('display.max_columns',75)
data
```

```python
import ast
data['genres']=data['genres'].map(lambda x: ast.literal_eval(x))
data['cast']=data["cast"].map(lambda x: ast.literal_eval(x))
data['crew']=data['crew'].map(lambda x: ast.literal_eval(x))
```

```python
def make_genresList(x):
    gen=[]
    st=" "
    for i in x:
        if i.get('name')=='Science Fiction':
            scifi='Sci-Fi'
            gen.append(scifi)
        else:
            gen.append(i.get('name'))
    if gen==[]:
        return np.NaN
    else:
        return (st.join(gen))
```

```python
data["genres_list"]=data['genres'].map(lambda x: make_genresList(x))
```

```python
data["genres_list"]
```

```python
def get_actor1(x):
    casts=[]
    for i in x:
        casts.append(i.get('name'))
    if casts==[]:
        return np.NaN
    else:
        return (casts[0])
```

```python
data["actor_1_name"]=data['cast'].map(lambda x: get_actor1(x))
```

```python
data["actor_1_name"]
```

```python
import numpy as np

def get_actor2(x):
    casts = [i.get('name') for i in x]  # Extract actor names

    if len(casts) < 2:  # Check if at least two actors exist
        return np.NaN
    else:
        return casts[1]  # Return the second actor


```

```python
data["actor_2_name"]=data['cast'].map(lambda x: get_actor2(x))
```

```python
data['actor_2_name']
```

```python
def get_actor3(x):
    casts = [i.get('name') for i in x]  # Extract actor names

    if len(casts) < 3:  # Check if at least two actors exist
        return np.NaN
    else:
        return casts[2] 
```

```python
data["actor_3_name"]=data['cast'].map(lambda x: get_actor3(x))
```

```python
def get_director(x):
    dt=[]
    st=' '
    for i in x:
        if i.get('job')=="Director":
            dt.append(i.get('name'))
    if dt==[]:
        return np.NaN
    else:
        return (st.join(dt))
```

```python
data['director_name']=data['crew'].map(lambda x: get_director(x))
```

```python
movie=data.loc[:,['director_name','actor_1_name','actor_2_name','actor_3_name','genres_list','title']]
movie
```
